```python
import pandas as pd
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, BaggingClassifier
from sklearn.metrics import f1_score
from sklearn.preprocessing import StandardScaler

train_df=pd.read_csv("finsecure_train.csv")
test_df=pd.read_csv("finsecure_test.csv")
train_df
```

```python
X_train=train_df.iloc[:, 1:31]
y_train=train_df["Status"]
X_train
y_train
X_test=test_df.iloc[:, 1:31]
```

```python
#Decision Tree
dt=DecisionTreeClassifier()
dt=dt.fit(X_train,y_train)
dt_pred=dt.predict(X_test)
results_dt=pd.DataFrame({
    "Index": test_df["Index"],
    "Status": dt_pred
})
results_dt.to_csv("submission_1_DT.csv", index=False)
```

```python
#KNN
knn=KNeighborsClassifier(n_neighbors=10)
knn.fit(X_train,y_train)
knn_pred=knn.predict(X_test)
results_knn=pd.DataFrame({
    "Index": test_df["Index"],
    "Status": knn_pred
})
results_knn.to_csv("submission_1_KNN.csv", index=False)
```

```python
#Naive Bayes
nb=GaussianNB()
nb.fit(X_train,y_train)
nb_pred=nb.predict(X_test)
results_nb=pd.DataFrame({
    "Index": test_df["Index"],
    "Status": nb_pred
})
results_nb.to_csv("submission_1_NB.csv", index=False)
```

```python
#SVM
svm=SVC()
svm.fit(X_train, y_train)
svm_pred=svm.predict(X_test)
results_svm=pd.DataFrame({
    "Index": test_df["Index"],
    "Status": svm_pred
})
results_svm.to_csv("submission_1_SVM.csv", index=False)
```

```python
#Ensemble - Random forest
rf=RandomForestClassifier()
rf.fit(X_train, y_train)
rf_pred=rf.predict(X_test)
results_rf=pd.DataFrame({
    "Index": test_df["Index"],
    "Status": rf_pred
})
results_rf.to_csv("submission_1_RF.csv", index=False)
```

```python
#Ensemble - AdaBoost
ada=AdaBoostClassifier(random_state=42)
ada.fit(X_train, y_train)
ada_pred=ada.predict(X_test)
results_ada=pd.DataFrame({
    "Index": test_df["Index"], 
    "Status": ada_pred
})
results_ada.to_csv("submission_1_ADA.csv", index=False)
```
